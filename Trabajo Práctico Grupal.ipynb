Exploracion de la estructura: dimensiones, columnas, tipos de dato y cobertura etaria.

```python
datos.info() #vemos si hay datos faltantes y características de las variables
print(f'El dataset contiene {datos.shape[0]} filas y {datos.shape[1]} columnas. Con variables numericas y categoricas de tipo float, int y obj. Ninguna fila tiene valores null.') #dimensiones dataframe
```

```python
print('Cantidad de edades distintas en la muestra:', datos['age'].nunique())  #Cuenta todas las edades diferentes que aparecen en la muestra
print(f'La covertura etaria es: {list(map(int, sorted(datos['age'].unique())))}') #Todas las edades que aparecen en la muestra en orden ascendete
```

```python
print(datos.isnull().sum()) #Cuenta la cantidad de valores nulos de cada varaibles
print('No se detectaron valores faltantes en las variables')
```

```python
print(f'La cantidad de filas duplicadas son: {datos.duplicated().sum()}') #Cuenta filas repetidas por completo, todos los valores coinciden con los de una fila anterior.
datos[datos.duplicated()]
ids_rep = int(datos['participant_id'].duplicated().sum()) #verificamos si hay identificadores repetidos
print(f'Hay {ids_rep} identificadores repetidos.')
print('Finalmente, no se identificaron registros completamente duplicados ni identificadores de participantes repetidos.')
```

```python
resumen_variables_num = datos.describe().transpose() #Evaluamos parámetros estadísticos principales de variables num
variables = ['gender','primary_platform','cognitive_decay_tier'] #variables que contienen únicamente valores obj
resumen_variables_obj = datos[variables].describe().T[['unique','top','freq']] #Evaluamos parámetros estadísticos principales para variables obj (unique,top,freq)
res_id = datos['participant_id'].describe()[['unique']] #parametro significativo para el identificador
display(resumen_variables_num,resumen_variables_obj,res_id)
```
